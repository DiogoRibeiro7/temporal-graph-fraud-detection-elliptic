# 06 — Temporal Validation and Leakage Control

Evaluate as a future-facing fraud system.


## Why this matters

Random splits can leak future information.


## Temporal split

Train on earlier time steps and test on later time steps.


## Decision metric

Use top-K metrics for analyst capacity.


```python
from graph_fraud.synthetic import make_synthetic_transaction_graph
from graph_fraud.validation import assert_no_temporal_leakage, temporal_train_test_split

nodes, _ = make_synthetic_transaction_graph(n_nodes=500)
train, test, cutoff = temporal_train_test_split(nodes, test_time_steps=3)
assert_no_temporal_leakage(train, test)
print(cutoff, len(train), len(test))
```
