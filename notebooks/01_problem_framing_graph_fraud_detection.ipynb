# 01 — Problem Framing: Graph Fraud Detection

Frame fraud detection as an investigation-support problem, not a generic classifier.


## Operational question

Can graph structure help prioritize suspicious Bitcoin transactions for analyst review?


## Why graph structure matters

Fraud risk can be relational. Neighbourhoods, flows, and connected components may carry useful signal.


## Success beyond accuracy

Use PR-AUC, precision@K, recall@K, temporal validation, and analyst capacity metrics.


## Caution

A score is a signal, not evidence. The model supports human review.
